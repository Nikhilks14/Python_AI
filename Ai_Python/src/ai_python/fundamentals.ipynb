```python
users = ["Nikhil" , "Amit" , "Jas" ]

print("For-each over a list:")
for user in users:
    print(f"    - {user}")

print("Clasic counting loop with range():")
for i in range(3,8,2):
    print(f"    i = {i}")


print("While loop : ")
retries = 0
while retries < 3:
    retries += 1
    print(f"    retry attempt {retries}")

print("continue / break :")
items = ["a", None, "b", "stop", "c"]
for item in items:
    if item is None:
        continue
    if item == "stop":
        break
    print(f"    processing {item}")
```

```python
a = [1,2]
b = [1,2]
c = a

print(a is b) # is checks the reference 
print(a==b)   # == checks the values 

print(f"a == c {a==c}")
print(f"a is c {a is c}")
```

```python
# list - ordered , mutable
users = ["Anuj" , "Rahul"]
users.append("Amit")
print(f"list -> {users}, first item -> {users[0]}")


#dict - key/value pairs
user = {"name": "Anuj" , "age" : 26}
print(f"dict -> {user['name']}")
print(f".get on missing key    -> {user.get('email')}")
print(f".get with default   -> {user.get('email', 'unknown')}")

for Key, value in user.items():
    print(f"    {Key} = {value}")


# tuple - ordered , immutable 
point = {10,20}
print(f"tuple -> {point}")


#Set - unordered , unique values only
tags = {"agent", "rag", "agent"}
print(f"set (duplicates collapsed ) -> {tags}")
```

```python
def greet(name):
    return f"Hello {name}"

greet("Anuj")


def greet1(name: str):
    return f"Hello {name}"

greet("Anuj")


def greet2(name) -> int:
    return f"Hello {name}"


greet1(12)
greet2("Anuj1")
```

```python
def greet(name: str, lastName = "Kumar"):
    return f"Hello {name} {lastName}"

greet("Anuj")
```

```python
def greet(name: str, age , lastName = "Kumar"):
    return f"Hello {name} {lastName}, age is {age}"

greet("Anuj",12)
```

```python
#Creating custom exceptions for API errors / validation
class APIValidationError(Exception):
    """Custom exception raised when API validation fails"""
    pass

def process_api_payload(payload: dict):
    if "prompt" not in payload:
        raise APIValidationError("Missing required 'prompt' key in payload")
    return f"Processing prompt: {payload['prompt']}"
```
