```python
class Dog:

    species = "Dog"

    def __init__(self, name: str , breed: str):
        self.name = name
        self.breed = breed
    
    def bark(self):
        self.eyes = 2
        return f"{self.name} says Woof! and has {self.eyes}"


    @classmethod
    def walk(cls):
        print(f"{cls.species} is walking")

Dog.walk()

leo = Dog(name ="Leo", breed = "Libra")

# print(leo.bark())
# print(leo.eyes)
# print(leo.breed)
```

### @ClassMethod and @StaticMethod

```python
def add_number(*args):
    print(type(args))
    print(args)

add_number(1,2,3,True)
```

```python
def create_profile(**kwargs):
    print(type(kwargs))
    print(kwargs)

create_profile(name ="Nikhil", age =30)
```

```python
def create_profile(**kwargs):
    print(type(kwargs))
    print(kwargs)

profile = {
    "name" : "Nikhil",
    "age" : 14
}

create_profile(**profile)
```

```python
def add(a,b,c):
    return sum((a,b,c))

k = [1,2,3]
print(add(*k))
```
