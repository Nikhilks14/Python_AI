Write a comment explaining each result.

```python
list = [ 0, 0.0, "", " ", "0", [], [0], None]

def about_type(Values: list) -> str:
    for item in Values:
        print(f"{item!r:5} ,{ type(item).__name__:5}, {bool(item)}")

if __name__ == "__main__":

    print(about_type(list))
```
