## 2. String Formatting & Manipulation

### Problem 2.1: Dynamic Prompt Generator

Build a string-processing function `generate_llm_prompt()` that formats incoming system instructions and dynamic user inputs for an AI API.

**Requirements:**

1. Accept three parameters: `system_role` (str), `user_query` (str), and `temperature` (float).
2. Clean `user_query` by removing leading/trailing whitespaces.
3. Use an **f-string** with specific formatting specs:
- Format `temperature` to exactly 2 decimal places (e.g., `0.70`).
- Convert `system_role` to uppercase.
1. Capitalize the first letter of `user_query`.

**Expected Input:**

```python
generate_llm_prompt("  system assistant  ", "   what is rag?   ", 0.7)
```

**Expected Output:**

```
[SYSTEM: ASSISTANT | TEMP: 0.70] What is rag?
```

````python
def generate_llm_prompt(system_role: str, user_query: str, temperature: float) -> str:
    # Clean whitespace
    role = system_role.strip().upper()
    query = user_query.strip()

    # The template already has a literal "SYSTEM:" label, so drop a leading
    # "SYSTEM" from the role to avoid "SYSTEM: SYSTEM ASSISTANT".
    role = role.removeprefix("SYSTEM").strip()

    # Capitalize first letter only (str.capitalize() would lowercase the rest,
    # e.g. "what is RAG?" -> "What is rag?")
    query = query[:1].upper() + query[1:]

    # f-string with format specs: :.2f -> exactly 2 decimals
    return f"[SYSTEM: {role} | TEMP: {temperature:.2f}] {query}"


if __name__ == "__main__":
    print(generate_llm_prompt("  system assistant  ", "   what is rag?   ", 0.7))
    # [SYSTEM: ASSISTANT | TEMP: 0.70] What is rag?

    print(generate_llm_prompt("assistant", "explain RAG", 1))
    # [SYSTEM: ASSISTANT | TEMP: 1.00] Explain RAG
````

### Problem 3.1: API Retry Counter with Exponential Backoff

Simulate an API request retry handler using a loop.

**Requirements:**

1. Given a list of simulated server responses: `responses = [500, 502, 503, 200, 404]`.
2. Iterate through `responses` using a `while` loop or `for` loop:
- If the code is `500`, `502`, or `503`, print a warning message indicating a retry, increment the attempt counter, and double the wait time (start backoff at 1 second).
- If the code is `200`, print success and `break` out of the loop immediately.
- For any other error (e.g., `404`), print a fatal error and `break`.
1. Skip `None` or invalid values in the list using `continue`.

````python
responses = [500, 502, 503, 200, 404]

for resp in responses:
    print(f"response" , resp)

def condition(status: str) -> str:
    while status == "500":
        print("indicating a retry")
        continue 

print(condition("500"))
````

````python
def simulate_retries(responses: list) -> None:
    attempts = 0
    wait = 1  # backoff starts at 1 second

    for code in responses:
        # Skip None / invalid (non-int) values. bool is a subclass of int, so exclude it.
        if code is None or not isinstance(code, int) or isinstance(code, bool):
            print(f"Skipping invalid response: {code!r}")
            continue

        if code in (500, 502, 503):
            attempts += 1
            print(f"Warning: server error {code}. Retry #{attempts} after {wait}s")
            wait *= 2  # exponential backoff: 1, 2, 4, ...
        elif code == 200:
            print(f"Success (200) after {attempts} retries")
            break
        else:
            print(f"Fatal error: {code}. Not retrying")
            break


if __name__ == "__main__":
    simulate_retries([500, 502, 503, 200, 404])
    # Warning: server error 500. Retry #1 after 1s
    # Warning: server error 502. Retry #2 after 2s
    # Warning: server error 503. Retry #3 after 4s
    # Success (200) after 3 retries

    simulate_retries([500, None, "bad", 404, 200])
    # Warning: server error 500. Retry #1 after 1s
    # Skipping invalid response: None
    # Skipping invalid response: 'bad'
    # Fatal error: 404. Not retrying
````

### Problem 4.1: Message Deduplication & Inverted Index

Write a script that processes a list of raw chat messages to deduplicate tags and index messages by user.

**Input Data:**

```python
chat_logs = [
    {"user": "Alice", "tags": ["python", "ai"], "msg": "Hello world"},
    {"user": "Bob", "tags": ["rag", "ai"], "msg": "Building a RAG pipeline"},
    {"user": "Alice", "tags": ["python"], "msg": "Learning collections"},
]
```

**Requirements:**

1. Create a `set` containing all **unique tags** across all chat messages.
2. Build a `dict` mapping each `user` to a `list` of their messages.
3. Access missing keys safely using `.get()` without raising a `KeyError`.

**Expected Outputs:**

```python
# Unique Tags Set:
{"python", "ai", "rag"}

# User Index Dict:
{
    "Alice": ["Hello world", "Learning collections"],
    "Bob": ["Building a RAG pipeline"]
}
```

````python
chat_logs = [
        {"user": "Bob", "tags": ["rag", "ai"], "msg": "Building a RAG pipeline"},
    {"user": "Alice", "tags": ["python", "ai"], "msg": "Hello world"},

    {"user": "Alice", "tags": ["python"], "msg": "Learning collections"},
]


def process_chat_logs(logs: list[dict]) -> tuple[set, dict]:
    unique_tags = set()
    user_index = {}

    for entry in logs:
        # 1. set.update() adds every tag; duplicates are dropped automatically
        unique_tags.update(entry["tags"])

        # 2. Build user -> messages. .get(key, []) returns a fresh empty list
        #    if the user is new, so no KeyError and no "if key in dict" check.
        user_index[entry["user"]] = user_index.get(entry["user"], []) + [entry["msg"]]

    return unique_tags, user_index


if __name__ == "__main__":
    tags, index = process_chat_logs(chat_logs)

    print(tags)   # {'python', 'ai', 'rag'} (set order is not guaranteed)
    print(index)
    # {'Alice': ['Hello world', 'Learning collections'],
    #  'Bob': ['Building a RAG pipeline']}

    # 3. Safe access with .get()
    print(index.get("Alice"))             # ['Hello world', 'Learning collections']
    print(index.get("Charlie"))           # None (no KeyError)
    print(index.get("Charlie", []))       # [] (custom default)
    # print(index["Charlie"])             # would raise KeyError
````
